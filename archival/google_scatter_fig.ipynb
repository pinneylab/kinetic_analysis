```python
import matplotlib.pyplot as plt
import random
```

```python
no_of_balls = 1000
x = [random.gauss(0.5, 0.1) for i in range(no_of_balls)]
y = [random.gauss(0.5, 0.1) for i in range(no_of_balls)]
# draw the plot
plt.figure()
plt.scatter(x, y, c='gray', alpha=0.85)
plt.axis([0.0, 1.0, 0.0, 1.0])
plt.xlabel("$K_m$")
plt.ylabel("$∆G_{fold}$")
plt.savefig("/Users/duncanmuir/Desktop/scatter_demo.svg", format="svg")
```
