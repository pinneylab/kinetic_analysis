# Interactive plotting example notebook
This notebook illustrates interactive plotting with a simple example.

## Simulate some data

First, we generate some example data for plotting.

```python
n = 100
x_data = [np.linspace(0, 100, 20)] * n

model = MichaelisMentenModel()
fits = []
y_data = []

for i in range(n):

    # sample kcat and km parameters
    params = {
        'kcat': np.random.normal(10, 1, size=1)[0],
        'km': np.random.normal(10, 1, size=1)[0]
        }
    
    # given parameters and independent variable array, calculate dependent variable array
    y = model(x_data[i], params) + np.random.normal(0, 0.1, x_data[i].shape)
    fits.append(params), y_data.append(y)
```

## Launch interactive plot

The `plot_data` function will initialize a server for the interactive plot. Navigate to the link on your browser to launch the application.

```python
# make some titles for each plot
# can include any information (i.e. fit parameters, etc)
titles = ['title_{}'.format(i) for i in range(n)]


plot_data(
    x_data, 
    y_data, 
    model=model, 
    fits=fits, 
    titles=titles,
    n_cols=3,
    n_rows=2,
    )
```
